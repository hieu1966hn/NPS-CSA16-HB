### Bài thực hành 1: 
mean: Tính tổng các phần tử sau đó chia cho số phần tử
median: Sắp xếp mảng và tìm phần tử ở giữa
mode: Phần tử xuất hiện nhiều lần nhất

```python
weights = pd.Series([37, 48, 42, 39, 100], index=["Hoa", "Nam", "Hải", 'Linh', 'Nhân']) 
print(weights.mean())
print(weights.median())
print(weights.mode())

```

### DATAFRAME
- DF > S
- Mỗi cột trong DF là 1 S

```python
animals = pd.DataFrame({ # giống với ctdl: dictinary
    "dong_vat": ["Sao la", 'Khỉ đuôi ngắn', 'Voi châu Á'],
    "tuoi_tho": [25, 25, 70],
    "loai": ["có vú", 'linh trưởng', 'có vú']
})

animals.info()
```

```python
animals.to_csv('./animals.csv',index=False) # lệnh tạo file CSV thông qua DF
```
