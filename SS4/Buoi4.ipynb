```python
import pandas as pd
df = pd.read_csv("2024 Billionaire List.csv", index_col=0)
df["2024 Net Worth"] = df["2024 Net Worth"].apply(lambda x : x[1 : -1]) # cắt bỏ ký tự đầu tiên $ và ký tự cuối B => giữ lại phần số
df["2024 Net Worth"] = pd.to_numeric(df["2024 Net Worth"]) # ép kiểu dữ liệu từ dạng obj/string => int/float
df.dtypes
df.head() # lấy 5 hàng đầu
```

```python
# Chuyển cột Age thành kiểu dữ liệu int
df["Age"] = df["Age"].astype('Int64')
df.dtypes
```

```python
# Lấy ra 2 cột Industry và 2024 Net Worth và nhóm các tỷ phú theo từng Industry => Chỉ tạo ra đối tượng, chưa hiển thị được kết quả tính toán
df[["Industry", "2024 Net Worth"]].groupby("Industry")
```

```python
# Tính tổng tài sản của tất cả các tỷ phú trong lĩnh vực đó
df[["Industry", "2024 Net Worth"]].groupby("Industry").sum()
```

```python
# Sắp xếp theo giá trị giảm dần
df_industry = df[["Industry", "2024 Net Worth"]].groupby("Industry").sum()
df_industry = df_industry.sort_values(by="2024 Net Worth", ascending=False) # giảm dần, ngược lại =True tăng dần
df_industry
```

```python
import pycountry_convert as pc # thư viện convert country to continent
def get_continent(country_name):
    try:
        country_code = pc.country_name_to_country_alpha2(country_name, cn_name_format="default")
        continent_code = pc.country_alpha2_to_continent_code(country_code)
        continent_name = pc.convert_continent_code_to_continent_name(continent_code)
        return continent_name
    except:
        return "Unknown"
    
df["Continent"] = df["Citizenship"].apply(get_continent)
df[["Continent", "2024 Net Worth"]].groupby("Continent").mean()
```
