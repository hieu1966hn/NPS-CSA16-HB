# Buổi 6: Trực quan hóa & Kể câu chuyện với dữ liệu

```python
# nhúng thư viện matplotlib vào đây
import matplotlib.pyplot as plt
import pandas as pd

```

```python
# Đọc file dữ liệu 
df = pd.read_csv('./spotify.csv')
df
```

```python
# xem kiểu dữ liệu từng cột
df.dtypes

# chuyển đổi KDL cột Date => kiểu ngày tháng "datetime64" bằng cách sử dụng hàm: 
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")
df.dtypes

# Làm sạch dữ liệu 
```

```python
# Sử dụng DataFrame.plot() để vẽ biểu đồ
df.plot.line(x="Date")
```
